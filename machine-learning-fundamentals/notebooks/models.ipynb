This notebook compares **regression** and **decision tree** models across two classic machine learning tasks:
1. **Regression**: predicting continuous values.
2. **Classification**: identifying categories.

## Part 1: Regression (House Price Prediction)

This experiment generates a synthetic dataset to predict property prices from features such as the number of bedrooms, property size, and building age.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score

# 1. Create a synthetic regression dataset
np.random.seed(42)
n_samples = 500

size_sqm = np.random.normal(150, 50, n_samples).clip(50, 400)
bedrooms = np.random.randint(1, 5, n_samples)
age_years = np.random.uniform(0, 30, n_samples)

# Generate prices from a linear relationship with added noise
price_brl = (size_sqm * 3000) + (bedrooms * 15000) - (age_years * 1000) + np.random.normal(0, 25000, n_samples)

regression_df = pd.DataFrame({
    'Size_sqm': size_sqm,
    'Bedrooms': bedrooms,
    'Age_years': age_years,
    'Price_BRL': price_brl
})

display(regression_df.head())
```

```python
# 2. Split the data into training and test sets
X_regression = regression_df[['Size_sqm', 'Bedrooms', 'Age_years']]
y_regression = regression_df['Price_BRL']

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_regression, y_regression, test_size=0.2, random_state=42)

# 3. Train the models
linear_regressor = LinearRegression()
linear_regressor.fit(X_train_reg, y_train_reg)

decision_tree_regressor = DecisionTreeRegressor(max_depth=5, random_state=42)
decision_tree_regressor.fit(X_train_reg, y_train_reg)

# 4. Generate predictions
y_pred_linear = linear_regressor.predict(X_test_reg)
y_pred_tree_reg = decision_tree_regressor.predict(X_test_reg)

# 5. Evaluate the models using MSE and R-squared
print("--- LINEAR REGRESSION ---")
print(f"MSE: {mean_squared_error(y_test_reg, y_pred_linear):.2f}")
print(f"R²: {r2_score(y_test_reg, y_pred_linear):.4f}")

print("\n--- DECISION TREE REGRESSOR ---")
print(f"MSE: {mean_squared_error(y_test_reg, y_pred_tree_reg):.2f}")
print(f"R²: {r2_score(y_test_reg, y_pred_tree_reg):.4f}")
```

## Part 2: Classification (Flower Species)

This experiment uses the classic **Iris** dataset to classify flower species with Logistic Regression and a Decision Tree Classifier.

```python
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Load the Iris dataset
iris = load_iris()
X_classification = pd.DataFrame(iris.data, columns=iris.feature_names)
y_classification = iris.target

# 2. Split the data into training and test sets
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X_classification, y_classification, test_size=0.2, random_state=42)

# 3. Train the models
logistic_regressor = LogisticRegression(max_iter=200)
logistic_regressor.fit(X_train_clf, y_train_clf)

decision_tree_classifier = DecisionTreeClassifier(max_depth=3, random_state=42)
decision_tree_classifier.fit(X_train_clf, y_train_clf)

# 4. Generate predictions
y_pred_logistic = logistic_regressor.predict(X_test_clf)
y_pred_tree_clf = decision_tree_classifier.predict(X_test_clf)

# 5. Evaluate the models
print(f"Accuracy - Logistic Regression: {accuracy_score(y_test_clf, y_pred_logistic):.4f}")
print(f"Accuracy - Decision Tree Classifier: {accuracy_score(y_test_clf, y_pred_tree_clf):.4f}")

print("\nDecision Tree Classification Report:")
print(classification_report(y_test_clf, y_pred_tree_clf, target_names=iris.target_names))
```
